```python
from pyspark.sql import SparkSession
import time
```

```python
spark = SparkSession.builder.master("local[1]") \
                    .appName('IntrotoSpark') \
                    .getOrCreate()
```

```python
# Load, show and process all data
!wget https://datasets.imdbws.com/title.akas.tsv.gz
!wget https://datasets.imdbws.com/title.basics.tsv.gz
```

```python
df = spark.read.format("csv").option("sep", "\t").option("header", "true").option("inferSchema", "true").load("title.akas.tsv.gz")
df_basics = spark.read.format("csv").option("sep", "\t").option("header", "true").option("inferSchema", "true").load("title.basics.tsv.gz")
```

```python
#Display the schema of a DataFrame
df.printSchema()

#Show the first 5 rows of the DataFrame
df.show(5)
```

```python
#Display the schema of a DataFrame
df_basics.printSchema()

#Show the first 5 rows of the DataFrame
df_basics.show(5)
```

```python
#Compute the number of titles per titleType
df_basics.groupBy("titleType").count().show()
```

```python
spark.stop()
```
